```python
import numpy as np

#сложение матриц

matrix1 = np.array([[1, 2, 4],
                    [7, 5, -2],
                    [-2, 1, 3]])
matrix2 = np.array([[-1, 0, 3],
                    [2, -6, 1],
                    [-7, 1, 5]])
print(matrix1 + matrix2)
```

```python
matrix1 = np.array([[0, 3, 4],
                    [-7, 1, -2],
                    [2, -1, 3]])
matrix2 = np.array([[1, 2, 4],
                    [7, 0, -2],
                    [-3, -2, 5]])
print(matrix1 + matrix2)
```

```python
#вычитание матриц

matrix1 = np.array([[5, 2, 9],
                    [7, 0, -2],
                    [-2, 8, 6]])
matrix2 = np.array([[1, -2, 1],
                    [7, 5, -2],
                    [-2, 1, 3]])
print(matrix2 - matrix1)
```

```python
# Умножение матриц

matrix1 = np.array([[1, 2, 4],
                    [7, 5, -2],
                    [-2, 1, 3]])
matrix2 = np.array([[-2, 3],
                    [4, 0],
                    [7, -1]])

result_matrix = np.dot(matrix1, matrix2)
print(result_matrix)
```

```python
#обратная матрица

matrix = np.array([
    [1, 2, 4],
    [7, 5, -2],
    [-2, 1, 3]
])

matrix_inv = np.linalg.inv(matrix)

print(matrix_inv)
```
