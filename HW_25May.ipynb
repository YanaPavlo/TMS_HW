```python
#Вывести все рейсы, совершенные из Москвы

SELECT *
FROM Trip
WHERE town_from = 'Moscow'
```

```python
# Вывести имена людей, которые заканчиваются на "man"

SELECT name
FROM Passenger
WHERE name LIKE '%man'
```

```python
# Вывести количество рейсов, совершенных на TU-134

SELECT count(id) as count
FROM Trip
WHERE plane = 'TU-134'
```

```python
# Какие компании совершали перелеты на Boeing

SELECT DISTINCT name
FROM Company
	INNER JOIN Trip ON Company.id = Trip.company
WHERE plane = 'Boeing'
```

```python
# Вывести все названия самолётов, на которых можно улететь в Москву 

SELECT DISTINCT plane
FROM Trip
WHERE town_to = 'Moscow'
```

```python
# Какие компании организуют перелеты из Владивостока

SELECT DISTINCT name
FROM Company
	INNER JOIN Trip on Company.id = Trip.company
WHERE town_from = 'Vladivostok'
```

```python
# Выведите идентификаторы всех рейсов и количество пассажиров на них. 
# Обратите внимание, что на каких-то рейсах пассажиров может не быть. В этом случае выведите число "0".

SELECT trip.id as id,
	COUNT(Pass_in_trip.Passenger) as count
FROM Trip
	LEFT JOIN Pass_in_trip ON Trip.id = Pass_in_trip.trip
GROUP BY Trip.id
```

```python
# В какие города летал Bruce Willis

SELECT DISTINCT town_to
FROM Trip
	INNER JOIN Pass_in_trip ON Trip.id = Pass_in_trip.trip
	INNER JOIN Passenger ON Pass_in_trip.passenger = Passenger.id
WHERE passenger.name = 'Bruce Willis'
```

```python
# Выведите идентификатор пассажира Стив Мартин и дату и время его прилёта в Лондон 

SELECT DISTINCT Passenger.id as id,
	time_in
FROM Trip
	INNER JOIN Pass_in_trip ON Trip.id = Pass_in_trip.trip
	INNER JOIN Passenger ON Pass_in_trip.passenger = Passenger.id
WHERE name = 'Steve Martin'
	AND town_to = 'London'
```

```python
# Определить, кто из членов семьи покупал картошку 

SELECT DISTINCT status
FROM FamilyMembers
WHERE FamilyMembers.member_id IN
    (
    SELECT family_member
    FROM Payments
    WHERE Payments.good IN
        (SELECT Goods.good_id
        FROM Goods
        WHERE good_name = 'potato')
```

```python
# Напишите запрос, который выведет имена пилотов, которые в качестве второго пилота в августе 2023 года летали в New York

SELECT DISTINCT name
FROM Pilots
WHERE pilot_id IN (
		SELECT second_pilot_id
		FROM Flights
		WHERE flight_date LIKE '2023-08-%'
			AND destination = 'New York'
	)
```

```python
# Найти имена всех матерей

SELECT member_name
FROM FamilyMembers
WHERE status = 'mother'
```

```python
# Сколько рейсов совершили авиакомпании из Ростова в Москву?

SELECT count(id) AS count
FROM Trip
WHERE town_from = 'Rostov'
	AND town_to = 'Moscow'
```

```python
# Сколько всего 10-ых классов

SELECT count(name) as count
FROM Class
WHERE name LIKE '10 %'
```

```python
# Выведите информацию об обучающихся, живущих на улице Пушкина (ul. Pushkina)?

SELECT *
FROM Student
where address LIKE 'ul. Pushkina%'
```

```python
# Сколько учениц с именем Анна (Anna) учится в школе?

SELECT count(first_name) AS count
FROM Student
WHERE first_name = 'Anna'
```

```python
# Сколько обучающихся в 10 B классе ?

SELECT COUNT(student) AS count
FROM Student_in_class
	INNER JOIN Class ON Student_in_class.class = Class.id
WHERE Class.name = '10 B'
```

```python
# Выясните, во сколько по расписанию начинается четвёртое занятие.

SELECT start_pair
FROM Timepair
WHERE id = '4'
```

```python
# Выведите идентификатор и признак наличия интернета в помещении. 
#Если интернет в сдаваемом жилье присутствует, то выведите «YES», иначе «NO».

SELECT id,
	IF(has_internet = '1', 'YES', 'NO') as has_internet
FROM Rooms
```

```python
# Выведите фамилию, имя и дату рождения студентов, кто был рожден в мае.

SELECT last_name,
	first_name,
	birthday
FROM Student
WHERE birthday LIKE '%-05-%'
```

```python
# Посчитай доход с женской аудитории (доход = сумма(price * items))

SELECT SUM(price * items) AS income_from_female
FROM Purchases
WHERE user_gender = 'f'
	OR user_gender = 'female'
```

```python
# Вывести список имён сотрудников, получающих большую заработную плату, чем у непосредственного руководителя

SELECT Subordinates.name
FROM Employee as Subordinates
	JOIN Employee as Chiefs ON Subordinates.chief_id = Chiefs.id
WHERE Subordinates.salary > Chiefs.salary
```

```python
# Выведите название страны, где находится город «Salzburg»

SELECT Countries.name AS country_name
FROM Countries
WHERE Countries.id = (
		SELECT countryid
		FROM Regions
		WHERE Regions.id = (
				SELECT regionid
				FROM Cities
				WHERE Cities.name = 'Salzburg'
			)
	)
```

```python
# Напишите запрос для поиска названия (name), населения (population) и площади (area) больших стран.

SELECT name,
	population,
	area
FROM World
WHERE area >= 3000000
	OR population >= 25000000
```
