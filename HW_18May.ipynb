```python
class Message:
    counter: int = 0

    @staticmethod
    def generate_id():
      Message.counter += 1
      return Message.counter

class Channel:
  def __init__(self, name: str):
    self.name = name

  def prepare_text(self, text: str):
    pass

  def send(self, text: str):
    send_text = self.prepare_text(text)
    return send_text

class SmsChannel(Channel):
  def __init__(self):
    super().__init__('SMS')

  def prepare_text(self, text: str):
    text = "SMS: " + text
    return text

class Logger:
  def __init__(self):
    self.loggs = []

  def add_to_log(self, id: int, channel_name: str, final_text:str):
    for_log = (f"ID: {id}, Channel:{channel_name}, Text: {final_text}")
    self.loggs.append(for_log)

  def show_log(self):
    for text in self.loggs:
      print (text)

my_logger = Logger()

sms = SmsChannel()

text_message = "I am doing HW"

current_id = Message.generate_id()

result = sms.send(text_message)

my_logger.add_to_log(current_id, sms.name, result)

my_logger.show_log()
```

```python
class Engine:
  def __init__(self, engine_type: str):
    self.engine_type = engine_type

class Car:

  @staticmethod
  def valid_vin(vin:str):
    if len(vin) == 17:
      return True
    else:
      return False

  def __init__(self, brand: str, vin: str, engine_type: str):
    if Car.valid_vin(vin) == False:
      engine_type = 'Incorrect VIN'

    self.brand = brand
    self.vin = vin
    self.engine = Engine(engine_type)

  def show_info(self):
    print(f"Brand: {self.brand}, VIN:{self.vin}, Engine type: {self.engine.engine_type}")

car1 = Car("VW", "VF3M45GXVHS197438", "Gasoline")
car1.show_info()

car2 = Car("BMW", "1234", "Electric")
car2.show_info()
```

```python
class TaxCalculator:
  def __init__(self, tax_rate: float):
    self.tax_rate = tax_rate

  def calculate_tax(self, amount: float):
    tax_amount = amount * self.tax_rate
    return tax_amount

class BasePayment:
  def __init__(self, percentage: float):
    self.percentage = percentage

  def calculate_comission (self, amount: float):
    comission = amount * self.percentage
    return comission

class BankCard(BasePayment):
  def __init__(self, percentage: float = 0.05):
    super().__init__(percentage)

class EWallet(BasePayment):
  def __init__(self, percentage: float = 0.02):
    super().__init__(percentage)

class PaymentProcessor:

  @staticmethod
  def valid_amount(amount: float):
    if amount > 0:
      return True
    else:
      return False

  def __init__(self, payment_method):
    self.payment_method = payment_method
    self.tax_calc = TaxCalculator(0.10)

  def process(self, amount: float):
    if PaymentProcessor.valid_amount(amount) == False:
      print('Incorrect amount')
      return
    else:
      comission = self.payment_method.calculate_comission(amount)
      tax = self.tax_calc.calculate_tax(amount)
      total = amount + comission + tax
      print(f"Amount: {amount}, Comission: {comission}, Tax: {tax}, Total: {total}")

my_card = BankCard()
processor1 = PaymentProcessor(my_card)
processor1.process(100.00)

my_wallet = EWallet()
processor2 = PaymentProcessor(my_wallet)
processor2.process(200.00)

processor1.process(-50)
```
